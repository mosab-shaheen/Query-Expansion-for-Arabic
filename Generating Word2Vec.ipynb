```python
import os
import re
import xml.etree.cElementTree as ET
from cltk.stop.arabic.stopword_filter import stopwords_filter
from cltk.corpus.arabic.alphabet import LETTERS
from nltk.stem.isri import ISRIStemmer
from gensim.models import word2vec
stemmer = ISRIStemmer()
nums='۰-۹'
letters=''.join(LETTERS)
reg='[^0-9A-Za-z'+str(nums)+str(letters)+']+'

def preprocess(text):
    txtList=stopwords_filter(text)
    l=[]
    for piece in txtList:
        piece=re.sub(reg, ' ', piece)
        piece=piece.split(' ')
        for word in piece:
            word=stemmer.stem(word)
            word=stopwords_filter(word)
            if len(word)!= 0:
                l.append(word[0])
    return l



tree = ET.parse("/home/mosab/Desktop/Link to Ass2/Ar/Arabic Corpus/clean.txt.utf8")
root = tree.getroot()
corpus=[]
for DOC in root.findall('DOC'):
    HEADLINE= DOC.find('HEADLINE').text.strip()
    TEXT= DOC.find('TEXT').text.strip()
    headlineTokens=preprocess(HEADLINE)
    corpus.append(headlineTokens)
    docTokens=preprocess(TEXT)
    corpus.append(docTokens)

model = word2vec.Word2Vec(corpus, size=100, window=6, min_count=1, workers=4)# default = 1 worker = no parallelization
model.wv.save_word2vec_format("/home/mosab/Desktop/IR/Ar/Project/word2vec.model.txt", binary=False)

print("==================================Embeddings are Generated==================================")

print(len(model.wv.vocab))
print(model.vector_size)

```
